```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
# 2. Combine into one master dataset
df_master = pd.read_csv("data_mingap.csv")

# 3. Create the target (Min_TS_all) 
# Ensure we use the exact same logic for every row
ts_cols = ['ΔG TS_endo_1', 'ΔG TS_endo_2', 'ΔG TS_exo_1', 'ΔG TS_exo_2']
df_master['target_raw'] = df_master[ts_cols].min(axis=1)

# 4. Perform a Random Split (80/20)
# This ensures bulky 'ID 5' molecules are in BOTH sets
train_df, test_df = train_test_split(df_master, test_size=0.20, random_state=42)

# 5. Calculate Master Scaling Constants from the NEW Train Set ONLY
MASTER_MEAN = train_df['target_raw'].mean()
MASTER_STD = train_df['target_raw'].std()

# 6. Apply Scaling
train_df['target_scaled'] = (train_df['target_raw'] - MASTER_MEAN) / MASTER_STD
test_df['target_scaled'] = (test_df['target_raw'] - MASTER_MEAN) / MASTER_STD

train_df.to_csv("data/train.csv")
test_df.to_csv("data.evaluate.csv")

stats = {
    'mean': MASTER_MEAN,
    'std': MASTER_STD
}
```
